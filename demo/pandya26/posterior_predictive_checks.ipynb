```python
from sapphire.utils import read_config
config = read_config.get({'path_config':'/mnt/ceph/users/vpandya/sapphire/scripts/config.yaml'})

### now modify for posterior sampling
config['runtype'] = 'sampling'
config['sampling_config']['method'] = 'posterior'
config['sampling_config']['posterior_file'] = 'numpyro_mix_chain%s_111_slope4_mzrgas.npz'
config['post_jit_benchmark'] = False

config['sampling_config']
```

```python
from sapphire.summaries import gaussian_kernel_regression as gkr

# copy-pasting return signature from module/function I'm calling 

(z0_Mvir, z0_smhm, fail_flag, Nfail, z0_Mstar, z0_fgas, z0_mzr,
 z0_Mdot_sfr,z0_Mdot_wind,z0_n0,z0_T0,z0_Mdot_in_halo,
 z0_Mdot_in_dm,z0_Mdot_cool,z0_mzr_gas, z0_halofb) = gkr.extract_posterior_quantities(sol)
```

```python
### also z=2 

(z2_Mvir, z2_smhm, fail_flag, Nfail, z2_Mstar, z2_fgas, z2_mzr,
 z2_Mdot_sfr,z2_Mdot_wind,z2_n0,z2_T0,z2_Mdot_in_halo,
 z2_Mdot_in_dm,z2_Mdot_cool,z2_mzr_gas, z2_halofb) = gkr.extract_posterior_quantities(sol,zindex=0)
```

```python
### we can extract other quantities manually following gkr.extract_posterior_quantities:
### for example, we'll plot tcool/tff ratio

tcool_eff = sol.ys[1]['tcool_eff'][:,:,-1] 
tff_eff = sol.ys[1]['tff_eff'][:,:,-1] 
```

```python
%matplotlib inline 

import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm, Normalize
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
from matplotlib.cm import ScalarMappable

# in case user loads module separately from sapphire.run()
from jax import config as jax_config
jax_config.update("jax_enable_x64", True) # required to accurately solve and take gradients through our diffeqs 

import jax
import jax.numpy as jnp
from astropy.table import Table,Column
from astropy.cosmology import Planck15
import os
import numpy as np # for matplotlib
```

```python
from sapphire.utils import read_obs
obs_mix = read_obs.read_mix({'data_path':'/mnt/ceph/users/vpandya/sapphire/data/','obs_name':'mix'}) # behroozi SMHM
```

```python
fig, axes = plt.subplots(nrows=3,ncols=3,figsize=(13,13),constrained_layout=True)
allax = np.ravel(axes)

# mark the z=2 panels w/ red outlines
for ax in allax[5:]:
    ax.spines['top'].set_edgecolor('red')
    ax.spines['right'].set_edgecolor('red')
    ax.spines['bottom'].set_edgecolor('red')
    ax.spines['left'].set_edgecolor('red')

# make borders thick
for ax in allax:
    for spine in ax.spines.values():
        spine.set_linewidth(3)

    ax.tick_params(axis='both',which='both',length=4, width=1.5)

def plot_nw(ax,x,y,fail_flag,color,zorder=0,label='__none__',cmap=plt.cm.coolwarm,norm=LogNorm(vmin=5e-2,vmax=1),alpha=0.2,lw=1):
    # adapted from gkr.summarize_mock
    
    bw, x0 = gkr.get_bandwidths(fail_flag, x)
    avg_y, err_y = gkr.nadaraya_watson(x, y, x0, bw)
    if type(color) is str:
        ax.plot(x0,avg_y,'-',color=color,alpha=alpha,lw=lw,label=label,zorder=zorder)
    else:
        ax.plot(x0,avg_y,'-',color=cmap(norm(color)),alpha=alpha,lw=lw,label=label,zorder=zorder)


###### z=0

# SFMS
tmanga = Table.read('/mnt/ceph/users/vpandya/sapphire_outputs/test_manga_sapphire_subsample.fits',format='fits')
allax[0].plot(tmanga['log_Mass'], tmanga['log_SFR_Ha'],'ms',ms=2,mec='none',alpha=0.2,label='MaNGA',zorder=np.inf)

allax[0].plot(z0_Mstar[0],jnp.log10(z0_Mdot_sfr[0]),'.',color='orange',mec='k',alpha=0.5,label='example realization')

for i in range(z0_Mstar.shape[0]):
    plot_nw(allax[0],z0_Mstar[i],jnp.log10(z0_Mdot_sfr[i]),fail_flag[i],'orange',zorder=np.inf,
            label='posterior regressions' if i==0 else '__none__')

leg = allax[0].legend(loc='best',fontsize=9,fancybox=True)

allax[0].set_xlabel(r'$\log M_*/M_{\odot}$',fontsize=16)
allax[0].set_ylabel(r'$\log$ SFR / [$M_{\odot}$/yr]',fontsize=16)

### mass loading at z=0 vs mstar (this is a prediction both because of mstar on x-axis and beyond MaNGA/Behroozi range)

for i in range(z0_Mstar.shape[0]):
    plot_nw(allax[1],z0_Mstar[i],z0_Mdot_wind[i]/z0_Mdot_sfr[i],fail_flag[i],'orange',zorder=np.inf)

allax[1].set_yscale('log')
allax[1].set_xlabel(r'$\log M_*/M_{\odot}$',fontsize=16)
allax[1].set_ylabel(r'$\eta_M\equiv\dot{M}_{\rm wind}/$SFR',fontsize=16)

### tcool/tff color-coded by A_E/A_M wind specific energy

wind_es = 10**free_params[:,2] / 10**free_params[:,0]
cmap = plt.cm.coolwarm
norm = LogNorm(vmin=0.05,vmax=1.0)

for i in range(z0_Mvir.shape[0]):
    plot_nw(allax[2],z0_Mvir[i],tcool_eff[i] / tff_eff[i],fail_flag[i],wind_es[i],
            cmap=cmap,norm=norm,alpha=0.5)

allax[2].set_yscale('log')

allax[2].axhline(10,color='b',ls='--',zorder=np.inf,lw=3)
allax[2].text(10.15,5,r'$t_{\rm cool}/t_{\rm ff}=10$'+'\nprecipitation limit',fontsize=12,color='b')

allax[2].set_xlabel(r'$\log M_{\rm vir}/M_{\odot}$',fontsize=16)
allax[2].set_ylabel(r'$\log t_{\rm cool}/t_{\rm ff}$',fontsize=16)

cax = inset_axes(allax[2],width="40%",height="5%",loc='lower left',bbox_to_anchor=(0.5,0.85,1,1),bbox_transform=allax[2].transAxes)
sm = ScalarMappable(norm=norm, cmap=cmap)
cbar = plt.colorbar(sm, cax=cax,orientation='horizontal')
cbar.set_label(r'$A_E/A_M$',fontsize=16)
# cbar.set_ticks([1e-2, 1e-1, 1])

### Halo and ISM inflow relative to DM inflow

for i in range(z0_Mvir.shape[0]):
    if i > 0:
        break  # for pandya+26, we don't let f_prev vary, so all realizations are same, so only plot one
    plot_nw(allax[3],z0_Mvir[i],z0_Mdot_in_halo[i]/z0_Mdot_in_dm[i]/0.158,fail_flag[i],'k',zorder=np.inf,lw=4,alpha=0.9,
            label='halo inflow' if i==0 else '__none__')

allax[3].axhline(1.0,ls='--',color='gray',zorder=np.inf,lw=3,label=r'$f_{\rm b}=0.158$ (Planck15)')

allax[3].set_xlabel(r'$\log M_{\rm vir}/M_{\odot}$',fontsize=16)
allax[3].set_ylabel(r'$\dot{M}/(f_{\rm b} \dot{M}_{\rm vir})$',fontsize=16)

for i in range(z0_Mvir.shape[0]):
    plot_nw(allax[3],z0_Mvir[i],z0_Mdot_cool[i]/z0_Mdot_in_dm[i]/0.158,fail_flag[i],wind_es[i],
            cmap=cmap,norm=norm,alpha=0.5,label='ISM inflow' if i==0 else '__none__')

for i in range(z0_Mvir.shape[0]):
    plot_nw(allax[3],z0_Mvir[i],z0_Mdot_sfr[i]/z0_Mdot_in_dm[i]/0.158,fail_flag[i],'orange',zorder=np.inf,
            alpha=0.2,label='SFR' if i==0 else '__none__')

allax[3].set_yscale('log')

leg = allax[3].legend(loc='lower right',fontsize=9,fancybox=True,ncol=1)
for l in leg.get_lines():
    l.set_alpha(1.0)

### halo f_b
# for i in range(z0_Mvir.shape[0]):
#     plot_nw(allax[5],z0_Mvir[i],z0_halofb[i],fail_flag[i],'orange',zorder=np.inf)

for i in range(z0_Mvir.shape[0]):
    plot_nw(allax[4],z0_Mvir[i],10**z0_halofb[i]/0.158,fail_flag[i],wind_es[i],zorder=np.inf,cmap=cmap,norm=norm,alpha=0.5)

allax[4].axhline(1.0,ls='--',color='gray',zorder=np.inf,lw=3)
allax[4].text(10.1,0.94,r'$f_{\rm b}=0.158$ (Planck15)',color='gray',fontsize=12)

allax[4].set_xlabel(r'$\log M_{\rm vir}/M_{\odot}$',fontsize=16)
allax[4].set_ylabel(r'Halo f$_{\rm b}$',fontsize=16)

############# z=2 

### SFMS 

### z=2 SFMS -- overplot as red lines in main SFMS panel
allax[5].plot(z2_Mstar[0],jnp.log10(z2_Mdot_sfr[0]),'.',color='orange',mec='k',alpha=0.2,label='__none__')

### z=2 SFMS -- overplot as red lines in main SFMS panel
for i in range(z2_Mstar.shape[0]):
    plot_nw(allax[5],z2_Mstar[i],jnp.log10(z2_Mdot_sfr[i]),fail_flag[i],'orange',zorder=np.inf)

# literature SFMS relations
# Whitaker+14 - broken power law 
w14_mstar_low, w14_mstar_high = np.linspace(9,10.2), np.linspace(10.2,11)
# w14_sfms = -19.99 + 3.44*mstar_arr - 0.13*mstar_arr**2 
w14_sfms_low = 0.91*(w14_mstar_low-10.2) + 1.62
w14_sfms_high = 0.67*(w14_mstar_high-10.2) + 1.62
allax[5].plot(w14_mstar_low,w14_sfms_low,'k-',label='Whitaker+14 ($z=2.0-2.5$)')
allax[5].plot(w14_mstar_high,w14_sfms_high,'k-',label='__none__')

# Speagle+14 meta-analysis combining many different studies onto common set of assumptions -- eqn 28 
z2_Gyr = Planck15.age(2).value # cosmic age in z=2 , will also be used for Pandya+17 below
s14_mstar = np.linspace(9.5,11)
s14_sfms = (0.84 - 0.026*z2_Gyr)*s14_mstar - (6.51 - 0.11*z2_Gyr)
allax[5].plot(s14_mstar,s14_sfms,'r--',label='Speagle+14 ($z=2$)')

# Pandya+17 - polynomial fit assuming zero-slope sSFR-Mstar relation
p17_mstar = np.linspace(9.0,11)
p17_sfms = (-0.0011*z2_Gyr**3 + 0.0233*z2_Gyr**2 - 0.2766*z2_Gyr - 7.8597) + p17_mstar # adding logmstar to go from logssfr to logsfr
allax[5].plot(p17_mstar,p17_sfms,'--',color='limegreen',zorder=np.inf,label='Pandya+17 ($z=2$)')

# JWST goes to lower mstar, this paper emphasizes systematics from dust (see also z~1.3 Neufeld+24)
c24_mstar = np.linspace(8,10.75)
c24_sfms = 0.95*(c24_mstar - 9.16) + 0.16
allax[5].plot(c24_mstar,c24_sfms,'--',color='cyan',zorder=np.inf,label='Clarke+24 ($z=1.4-2.7$)')


leg = allax[5].legend(loc='best',fontsize=9,fancybox=True)

allax[5].set_xlabel(r'$\log M_*/M_{\odot}$',fontsize=16)
allax[5].set_ylabel(r'$\log$ SFR / [$M_{\odot}$/yr] (z=2)',fontsize=16)



### SMHM
for i in range(z2_Mvir.shape[0]):
    plot_nw(allax[6],z2_Mvir[i],z2_smhm[i],fail_flag[i],'orange',zorder=np.inf,label='__none__')

allax[6].set_xlabel(r'$\log M_{\rm vir}/M_{\odot}$',fontsize=16)
allax[6].set_ylabel(r'$\log M_*/M_{\rm vir}$ (z=2)',fontsize=16)

# SMHM at z~2: behroozi+19
tum = Table.read(os.path.join(config['data_path'],'obs/lit/smhm_a0.334060.dat'),format='ascii')
tum = tum[(tum['HM(0)']>=10.0) & (tum['HM(0)']<=11.6)]    

allax[6].errorbar(tum['HM(0)'],tum['Med_Cen_SF(7)'],yerr=tum['Err-(9)'], # use lower error which is larger than upper
                  fmt='-',color='k',capsize=3,zorder=np.inf,label=r'Behroozi+19 ($z\sim2$)')

# SMHM at z~2 from Shuntov+24
ts24 = Table.read(os.path.join(config['data_path'],'obs/lit/shuntov24_SHMR.ecsv'))
ts24 = ts24[ts24['Redshift']=='2.0 < z < 2.5']
ts24 = ts24[ts24['M_halo']<10**11.6]
allax[6].errorbar(np.log10(ts24['M_halo']),np.log10(ts24['M_star_50']/ts24['M_halo']),
                  fmt='-',color='purple',capsize=3,zorder=np.inf,label=r'Shuntov+25 ($z=2.0-2.5$)')
allax[6].fill_between(np.log10(ts24['M_halo']),np.log10(ts24['M_star_16']/ts24['M_halo']),np.log10(ts24['M_star_84']/ts24['M_halo']),
                alpha=0.2,color='purple',zorder=0)

allax[6].legend(loc='best',fontsize=9,fancybox=True)

### fgas
for i in range(z2_Mstar.shape[0]):
    plot_nw(allax[7],z2_Mstar[i],z2_fgas[i],fail_flag[i],'orange',zorder=np.inf)

allax[7].set_xlabel(r'$\log M_*/M_{\odot}$',fontsize=16)
allax[7].set_ylabel(r'$\log M_{\rm ISM}/M_*$ (z=2)',fontsize=16)

### mzr_gas
for i in range(z2_Mstar.shape[0]):
    plot_nw(allax[8],z2_Mstar[i],z2_mzr_gas[i],fail_flag[i],'orange',zorder=np.inf,label='__none__')

# literature relations
OH_sun = 8.69
s21_mstar = np.linspace(9.25,10.25)
s21_mzr = 0.3*(s21_mstar-10.0) + 8.51 - OH_sun  # Sanders+21 MOSDEF 
allax[8].plot(s21_mstar,s21_mzr,'k:',label=r'Sanders+21 ($z\sim2.3$)')

h24_mstar = [7.42,8.20,9.09] # He+24 (GLASS)
h24_mzr = np.array([8.0,8.15,8.37]) - OH_sun
allax[8].plot(h24_mstar,h24_mzr,'rs-',label=r'He+24 (lensed $z=1.8-2.3$)')

k25_mstar = np.linspace(8.5,10.0)
k25_mzr = 8.44 + 0.27*(k25_mstar-10.0) - OH_sun # Khostovan+25 JWST z~2 AURORA
allax[8].plot(k25_mstar,k25_mzr,'m-',label=r'Khostovan+25 ($z\sim2.3$)')

allax[8].set_xlabel(r'$\log M_*/M_{\odot}$',fontsize=16)
allax[8].set_ylabel(r'$\log Z_{\rm ISM}/Z_{\odot}$ (z=2)',fontsize=16)

allax[8].legend(loc='lower right',fontsize=9,fancybox=True)

plt.savefig(os.path.join(config['output_path'],'figures/posterior_predictives_extras.png'),
            bbox_inches='tight',facecolor='w',dpi=300)
```
